```python
import numpy as np
np.random.seed(7)
cities = np.array(["Kathmandu", "Pokhara", "Biratnagar", "Lalitpur", "Birgunj"])
days = np.array([f"D{i:02d}" for i in range(1, 15)])
temps = np.random.randint(14, 36, size=(5, 14)).astype(float) # °C, 5 cities × 14 days
humidity = np.random.randint(40, 95, size=(5, 14)).astype(float) # %
temps[2, 5] = np.nan # sensor outage
temps[4, 11] = np.nan
```

## B1 — Array anatomy and reshaping (3 pts)

Print `temps.shape` , `.ndim` , `.dtype` , `.size` . Then build `np.arange(70)` , reshape it to the same shape as `temps` two ways — with explicit dimensions and with `-1` — show the two results are identical, and flatten it back. In a comment, explain what `-1` asked NumPy to work out, and why reshaping does not copy the data.

```python
print("B1 - temps shape:", temps.shape)
print("B1 - temps dimensions:", temps.ndim)
print("B1 - temps dtype:", temps.dtype)
print("B1 - temps size:", temps.size)

arr = np.arange(70)

reshaped_explicit = arr.reshape(5, 14)
reshaped_auto = arr.reshape(-1, 14)

print("B1 — Explicit reshape:")
print(reshaped_explicit)

print("B1 — Reshape using -1:")
print(reshaped_auto)

print("B1 — Are both reshapes identical:", np.array_equal(
    reshaped_explicit, reshaped_auto
))

flattened = reshaped_explicit.flatten()

print("B1 — Flattened array:")
print(flattened)

# -1 tells NumPy to automatically calculate the missing dimension based on
# the original size and the provided dimensions.
# Reshaping usually does not copy data because NumPy creates a new view that
# changes the way the same memory is interpreted.
```

## B3 — Boolean masks (3 pts)

(c) In a comment, explain in one sentence what a mask is, and why `temps[temps > 30]` cannot keep the 2D shape.

```python
hot_mask = temps > 30

count_hot_per_city = np.sum(hot_mask, axis=1)

print("B3a — Readings above 30 per city:",
      count_hot_per_city)

kathmandu_hot_days = days[hot_mask[0]]

print("B3b — Kathmandu days above 30:",
      kathmandu_hot_days)

# A NumPy mask is a True/False filter.
# It picks elements that satisfy a condition, e.g., temps > 30, and can be used to index an array.
# The result becomes a 1D array because the number of matching elements can vary, so a fixed multidimensional shape cannot be maintained.
# Internally, Boolean indexing creates a new array containing the selected elements in iteration order.
```

```python
# Taking a slice (view)
city_slice = temps[0, :3]

city_slice[0] = 999

print("B4 — Original array after modifying view:")
print(temps[0])

# Using copy()
city_copy = temps[1, :3].copy()

city_copy[0] = 999

print("B4 — Original array after modifying copy:")
print(temps[1])

# NumPy defaults to views because they are faster and memory efficient since
# they avoid creating unnecessary copies of large arrays.
# Forgetting this can cause accidental changes to the original data when we
# modify a sliced array.
```

## B5 — Vectorization vs. a Python loop (4 pts)

Convert `temps` to Fahrenheit two ways: with an explicit loop (the only loop allowed in Part B) and with one vectorized expression. Confirm they agree with `np.allclose` . Then time both on an array of 1,000,000 values using `time.perf_counter()` , and print both timings plus the speedup as a multiple ( ~Nx faster ). In a comment, explain where the loop's extra time actually goes.

```python
import time

# Explicit Python loop  

fahrenheit_loop = np.empty_like(temps)

for i in range(temps.shape[0]):
    for j in range(temps.shape[1]):
        fahrenheit_loop[i, j] = (temps[i, j] * 9/5) + 32


# Vectorized NumPy method

fahrenheit_vectorized = (temps * 9/5) + 32


print("B5 — Loop and vectorized results agree:",
      np.allclose(fahrenheit_loop, fahrenheit_vectorized))


# Timing comparison on 1,000,000 values

large_array = np.random.randint(14, 36, size=1000000).astype(float)

start = time.perf_counter()

large_loop_result = np.empty_like(large_array)

for i in range(len(large_array)):
    large_loop_result[i] = (large_array[i] * 9/5) + 32

loop_time = time.perf_counter() - start


start = time.perf_counter()

large_vectorized_result = (large_array * 9/5) + 32

vectorized_time = time.perf_counter() - start


speedup = loop_time / vectorized_time

print("B5 - Loop timing:", loop_time, "seconds")
print("B5 - Vectorized timing:", vectorized_time, "seconds")
print("B5 - Vectorization speedup:", speedup, "x faster")


# The loop spends extra time repeatedly executing Python-level operations,
# indexing, and interpreter overhead for every element, while NumPy performs
# calculations using optimized compiled C operations internally.
```

## B6 — Broadcasting (4 pts)

(a) Compute each city's own fortnight average with a NaN-aware function and `keepdims=True` , then subtract it from `temps` to get an anomaly grid; print the result's shape and confirm it is still `(5, 14)` . (b) Apply a per-day sensor correction of shape `(14,)` to all five cities in one expression. (c) Attempt `temps + np.array([1, 2, 3, 4, 5])` , catch the `ValueError` , print its message, and explain in a comment why aligning shapes from the right makes this fail while (b) succeeds.

```python
# (a) City-wise fortnight average

city_average = np.nanmean(temps, axis=1, keepdims=True)

anomaly_grid = temps - city_average

print("B6a - Anomaly grid:")
print(anomaly_grid)

print("B6a - Anomaly grid shape:",
      anomaly_grid.shape)

print("B6a - Shape confirmation:",
      anomaly_grid.shape == (5, 14))


# (b) Per-day sensor correction

sensor_correction = np.array(
    [0.5, -0.2, 0.1, 0.3, -0.4, 0.2, 0.6,
     -0.1, 0.4, -0.3, 0.5, 0.2, -0.2, 0.1]
)

corrected_temps = temps + sensor_correction

print("B6b - Corrected temperature grid shape:",
      corrected_temps.shape)


# (c) Broadcasting error example

try:
    result = temps + np.array([1, 2, 3, 4, 5])

except ValueError as error:
    print("B6c - ValueError message:")
    print(error)


# Broadcasting compares shapes from the right side.
# In (b), (5,14) and (14,) match because 14 aligns with 14 columns.
# In (c), (5,14) and (5,) fail because 14 cannot align with 5.
```

## B7 — Matrix multiplication (4 pts)

(a) On the grocery-basket example ( `quantities = [2, 3, 1]` , `prices = [120, 90, 250]` ), print both `*` and `@` , and state in a comment what each returned and why they differ. (b) Build `features` of shape `(14, 2)` whose columns are the across-city NaN-aware mean temperature and mean humidity per day, define `weights = np.array([0.6, -0.1])` , and compute a comfort score per day with `@` . Print the shapes of `features` , `weights` , and the result, and explain in a comment why the inner dimensions had to match.

```python
# (a) Grocery basket example

quantities = np.array([2, 3, 1])
prices = np.array([120, 90, 250])

print("B7a — Element-wise multiplication (*):",
      quantities * prices)

print("B7a — Matrix multiplication (@):",
      quantities @ prices)


# * performs element-by-element multiplication and returns an array.
# @ performs dot product multiplication and returns a single scalar value
# because it sums the products of corresponding elements.


# (b) Comfort score calculation

mean_temp_per_day = np.nanmean(temps, axis=0)
mean_humidity_per_day = np.nanmean(humidity, axis=0)

features = np.column_stack(
    (mean_temp_per_day, mean_humidity_per_day)
)

weights = np.array([0.6, -0.1])

comfort_score = features @ weights


print("B7b — Features shape:",
      features.shape)

print("B7b — Weights shape:",
      weights.shape)

print("B7b — Comfort score shape:",
      comfort_score.shape)


# The inner dimensions must match because matrix multiplication requires the
# number of columns in the first array to equal the number of rows in the second.
# Here features has shape (14,2) and weights has shape (2,), so multiplication
# is valid.
```

## B8 — Aggregating along an axis (4 pts)

Using NaN-aware aggregations, print: mean, max, min and std per city; mean and max per day; and each city's hottest day as a day label (use `argmax` , then index `days` ). State each result's shape. In a comment, state what `axis=0` and `axis=1` collapse here, in terms of this dataset.

```python
# Per city statistics

city_mean = np.nanmean(temps, axis=1)
city_max = np.nanmax(temps, axis=1)
city_min = np.nanmin(temps, axis=1)
city_std = np.nanstd(temps, axis=1)

print("B8 — Mean temperature per city:",
      city_mean,
      "shape:", city_mean.shape)

print("B8 — Maximum temperature per city:",
      city_max,
      "shape:", city_max.shape)

print("B8 — Minimum temperature per city:",
      city_min,
      "shape:", city_min.shape)

print("B8 — Standard deviation per city:",
      city_std,
      "shape:", city_std.shape)



# Per day statistics

day_mean = np.nanmean(temps, axis=0)
day_max = np.nanmax(temps, axis=0)

print("B8 — Mean temperature per day:",
      day_mean,
      "shape:", day_mean.shape)

print("B8 — Maximum temperature per day:",
      day_max,
      "shape:", day_max.shape)



# Hottest day for each city

hottest_day_index = np.nanargmax(temps, axis=1)

hottest_days = days[hottest_day_index]

print("B8 — Hottest day label for each city:",
      hottest_days,
      "shape:", hottest_days.shape)


# axis=0 collapses rows (cities), producing values for each day.
# axis=1 collapses columns (days), producing values for each city.
```

## B9 — Reproducible randomness and a train/test split (3 pts)

Flatten `temps` to 70 samples. With `np.random.seed(42)` , shuffle an index array and split it 80/20 into train and test index sets; print both shapes, confirm they sum to 70, and confirm they share no index. Re-seed with 42, repeat, and assert the split is identical. In a comment, explain why you shuffle before splitting, and why the seed matters when you share results.

```python
samples = temps.flatten()

indices = np.arange(samples.size)

np.random.seed(42)

np.random.shuffle(indices)

train_indices = indices[:56]
test_indices = indices[56:]


print("B9 - Training index shape:",
      train_indices.shape)

print("B9 - Testing index shape:",
      test_indices.shape)

print("B9 - Total samples:",
      train_indices.size + test_indices.size)

print("B9 - Shared indexes:",
      np.intersect1d(train_indices, test_indices).size == 0)



# Repeat with the same seed

np.random.seed(42)

indices_repeat = np.arange(samples.size)

np.random.shuffle(indices_repeat)

train_repeat = indices_repeat[:56]
test_repeat = indices_repeat[56:]


assert np.array_equal(train_indices, train_repeat)
assert np.array_equal(test_indices, test_repeat)

print("B9 - Split identical after reseeding:",
      True)


# Shuffling before splitting prevents the train and test sets from containing
# biased groups of ordered data.
# The seed ensures the random process produces the same split when results are
# shared or experiments are repeated.
```

## B10 — np.where , argmax , argmin (3 pts)

(a) Label every reading "hot" / "mild" / "cold" (above 30, 22–30, below 22) with nested `np.where` , and print the label grid's shape. (b) Given `scores = np.array([0.08, 0.62, 0.21, 0.09])` and `labels = ["rain", "clear", "cloudy", "fog"]` , print the predicted label using `argmax` . In a comment, explain why a classifier needs `argmax` rather than `max`

```python
# (a) Temperature labels

temperature_labels = np.where(
    temps > 30,
    "hot",
    np.where(
        temps >= 22,
        "mild",
        "cold"
    )
)

print("B10a - Temperature label grid:")
print(temperature_labels)

print("B10a - Label grid shape:",
      temperature_labels.shape)



# (b) Prediction using argmax

scores = np.array([0.08, 0.62, 0.21, 0.09])

labels = np.array(["rain", "clear", "cloudy", "fog"])

predicted_label = labels[np.argmax(scores)]

print("B10b - Predicted label:",
      predicted_label)


# A classifier needs argmax because it returns the position of the highest
# probability score, which allows us to select the corresponding class label.
# max only returns the highest value and loses the class information.
```
