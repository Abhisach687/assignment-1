## C10 — Derived columns (3 pts)

Add `passed` (`marks >= 40`), and `grade` using nested `np.where`: 85+ → A, 70–84 → B, 40–69 → C, below 40 → F. Print the overall pass rate as a percentage by taking the mean of `passed` cast to `int`, and print `grade.value_counts()`. Deal with the impossible marks values you found in C2, and say in a comment what you did with them and why.

```python
invalid_marks = ~df["marks"].between(0, 100)

print("Impossible marks:\n", df.loc[invalid_marks, ["student_id", "subject", "marks"]])
print("Rows before removing impossible marks:", len(df))

# Assume marks are out of 100. Remove values outside 0–100 before
# assigning grades or calculating summaries. We cannot infer the
# true values, and capping 150 at 100 would invent a perfect score.
# Keep raw_df unchanged so the removed records can still be audited.
df = df.loc[~invalid_marks].copy()

print("Invalid rows removed:", invalid_marks.sum())
print("Rows after removal:", len(df))


df["passed"] = df["marks"] >= 40
df["grade"] = np.where(
    df["marks"] >= 85,
    "A",
    np.where(
        df["marks"] >= 70,
        "B",
        np.where(df["marks"] >= 40, "C", "F"),
    ),
)

pass_rate = df["passed"].astype(int).mean() * 100

print("\nOverall pass rate: {:.2f}%".format(pass_rate))
print("\nGrade counts:\n", df["grade"].value_counts())
print("\nExample derived columns:\n", df[["marks", "passed", "grade"]].head())

# This is the pass rate across retained student-subject results,
# not the proportion of distinct students who passed every subject.
```

## C11 — Types and memory (3 pts)

Convert `subject` and `city` to the `category` dtype. Print `memory_usage(deep=True)` for those two columns before and after, in bytes, and the percentage saved. In a comment, explain what makes `category` cheaper, and when it would not help.


```python
category_columns = ["subject", "city"]

# Exclude the unchanged row index to measure only these columns.
memory_before = df[category_columns].memory_usage(index=False, deep=True)

df["subject"] = df["subject"].astype("category")
df["city"] = df["city"].astype("category")

memory_after = df[category_columns].memory_usage(index=False, deep=True)
percent_saved = (memory_before - memory_after) / memory_before * 100

print("Memory before (bytes):\n", memory_before)
print("\nMemory after (bytes):\n", memory_after)
print("\nPercentage saved per column:\n", percent_saved.round(2))

total_saved = (1 - memory_after.sum() / memory_before.sum()) * 100
print("\nTotal percentage saved: {:.2f}%".format(total_saved))
print("\nNew dtypes:\n", df[category_columns].dtypes)

# A category stores each distinct label once plus small integer codes
# for the rows. This helps because many rows repeat the same few cities
# and subjects. With almost all unique labels, the category lookup and
# codes can cost as much as, or more than, the original column.
# Exact byte counts depend on pandas and the Python environment.
```

## +5 — A reporting pipeline.

Wrap Part C into three functions — `load(path)`, `clean(df)`, `summarise(df)` — so that `summarise(clean(load("data/scores_raw.csv")))` reproduces your C12(a) table from the raw file in one line. It must run top to bottom with no manual steps.


```python
def load(path):
    """Read the CSV; resolve relative paths from the project folder."""
    path = Path(path)

    if not path.is_absolute():
        path = PROJECT_ROOT / path

    return pd.read_csv(path)


def clean(df):
    """Repeat the C5–C11 cleaning choices without changing the input."""
    result = df.copy(deep=True)

    result["city"] = result["city"].str.strip().str.title()

    # Match C6 exactly: compute the mean before dropping any rows.
    attendance_mean = result["attendance_percent"].mean()
    result["attendance_percent"] = result["attendance_percent"].fillna(
        attendance_mean
    )

    result = result.dropna(subset=["marks"]).copy()
    result = result.drop_duplicates(keep="first").copy()

    result["exam_date"] = pd.to_datetime(
        result["exam_date"], format="%Y-%m-%d"
    )
    result["exam_month"] = result["exam_date"].dt.to_period("M")
    result["exam_weekday"] = result["exam_date"].dt.day_name()

    result = result.loc[result["marks"].between(0, 100)].copy()

    result["passed"] = result["marks"] >= 40
    result["grade"] = np.where(
        result["marks"] >= 85,
        "A",
        np.where(
            result["marks"] >= 70,
            "B",
            np.where(result["marks"] >= 40, "C", "F"),
        ),
    )

    result["subject"] = result["subject"].astype("category")
    result["city"] = result["city"].astype("category")

    return result.sort_index()


def summarise(df):
    """Return the C12(a) table in a fixed subject order."""
    return (
        df.groupby("subject", observed=True)["marks"]
        .agg(["mean", "max", "count"])
        .sort_index()
    )


pipeline_summary = summarise(clean(load("data/scores_raw.csv")))

print(pipeline_summary)
print("\nEquals C12(a):", pipeline_summary.equals(subject_summary))

pd.testing.assert_frame_equal(pipeline_summary, subject_summary)
pd.testing.assert_frame_equal(clean(raw_df), df)

print("Pipeline summary and cleaned data both match the earlier answers.")
```
